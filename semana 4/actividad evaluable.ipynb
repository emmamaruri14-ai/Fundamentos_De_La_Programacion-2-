Descripción del programa:

Este programa genera e imprime una tabla de Pitágoras del 1 al 10 utilizando ciclos `for` y listas. Permite consultar el resultado de la multiplicación entre dos factores, validando que los valores ingresados sean números enteros dentro del rango permitido (1 al 10).

```python

def generar_tabla(n=10):
    tabla = []

    for i in range(1, n + 1):
        fila = []

        for j in range(1, n + 1):
            fila.append(i * j)

        tabla.append(fila)

    return tabla


def imprimir_tabla(tabla):
    print("TABLA DE PITÁGORAS")
    print("     ")

    for columna in range(1, 11):
        print(f"{columna:>5}", end="")

    print("\n" + "-" * 55)

    for i, fila in enumerate(tabla, start=1):
        print(f"{i:>3} |", end="")

        for valor in fila:
            print(f"{valor:>5}", end="")

        print()


def consultar_producto(tabla, renglon, columna):
    return tabla[renglon - 1][columna - 1]


def main():
    tabla = generar_tabla()
    imprimir_tabla(tabla)

    while True:
        print("\nCONSULTAR PRODUCTO")

        # Validar renglón
        while True:
            try:
                renglon = int(input("Ingresa el renglón (1-10): "))

                if renglon < 1 or renglon > 10:
                    print("Error: el renglón debe estar entre 1 y 10.")
                else:
                    break

            except ValueError:
                print("Error: debes ingresar un número entero.")

        # Validar columna
        while True:
            try:
                columna = int(input("Ingresa la columna (1-10): "))

                if columna < 1 or columna > 10:
                    print("Error: la columna debe estar entre 1 y 10.")
                else:
                    break

            except ValueError:
                print("Error: debes ingresar un número entero.")

        resultado = consultar_producto(tabla, renglon, columna)

        print(f"\n{renglon} x {columna} = {resultado}")

        continuar = input(
            "\n¿Deseas consultar otro producto? (si/no): "
        ).lower()

        if continuar != "si":
            print("Utilizaste la tabla de Pitágoras.")
            break


main()

```
