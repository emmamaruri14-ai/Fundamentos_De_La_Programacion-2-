```python
#ejercicio 1#

dias = ("Lunes", "Martes", "Miércoles", "Jueves", "Viernes", "Sábado", "Domingo")
km = (120, 85, 0, 210, 95, 0, 150)

dias_rodados = [dias[i] for i in range(len(km)) if km[i] > 0]

def resumen( km):
    total_km = sum(km)
    dias_con_km = sum(1 for x in km if x > 0)
    promedio_km = total_km / len(dias_rodados)
    dia_con_mas_km = dias[km.index(max(km))]
    return f"Total de km recorridos: {total_km}\nDías con km recorridos: {dias_con_km}\nPromedio de km por día: {promedio_km:.2f}\nDía con más km : {dia_con_mas_km}"

print(resumen(km))
```

```python
#ejercicio 3#
paradas = (("Querétaro", 0), ("San Juan del Río", 52),
           ("Tequisquiapan", 74), ("Bernal", 125), ("Jalpan", 260))

mayor_distancia = -1
tramo_mayor = ("", "")

for i in range(len(paradas) - 1):
    origen, km_origen = paradas[i]
    destino, km_destino = paradas[i + 1]

    distancia = km_destino - km_origen

    print(f"{origen} → {destino}: {distancia} km")

    if distancia > mayor_distancia:
        mayor_distancia = distancia
        tramo_mayor = (origen, destino)

print(f"Tramo más largo: {tramo_mayor[0]} → {tramo_mayor[1]} ({mayor_distancia} km)")
```

```python
#ejercicio 4#

temps = (8, 14, 22, 27, 31, 19, 11)

def clasificar(t):
    if t < 12:
        return "frío"
    elif t <= 24:
        return "templado"
    else:
        return "calor"

etiquetas = []
contador = {"frío": 0, "templado": 0, "calor": 0}

for temperatura in temps:
    etiqueta = clasificar(temperatura)
    etiquetas.append(etiqueta)
    contador[etiqueta] += 1

predominante = ""
mayor = -1

for etiqueta, cantidad in contador.items():
    if cantidad > mayor:
        mayor = cantidad
        predominante = etiqueta

print(etiquetas)
print(contador)
print(f"Predomina: {predominante}")
```

```python
#ejercicio 5#
setlist = (("Intro", 95), ("Riff asesino", 245),
           ("Balada", 302), ("Solo", 128), ("Encore", 409))

def formato(seg):
    minutos = seg // 60
    segundos = seg % 60
    return f"{minutos}:{segundos:02d}"

total = 0

for cancion, duracion in setlist:
    print(f"{cancion}: {formato(duracion)}")
    total += duracion

print(f"Duración total: {formato(total)}")
```

```python
#ejercicio 7#
notas = {
    "Ana": [90, 85, 78],
    "Luis": [60, 72, 65],
    "Mía": [100, 95, 98],
    "Beto": [70, 68, 74]
}

def promedio(lista):
    suma = 0
    for nota in lista:
        suma += nota
    return round(suma / len(lista), 1)

promedios = {}

for alumno, calificaciones in notas.items():
    prom = promedio(calificaciones)
    promedios[alumno] = prom

    estado = "Aprobado" if prom >= 70 else "Reprobado"
    print(f"{alumno}: {prom} - {estado}")

mejor = ""
mayor = -1

for alumno, prom in promedios.items():
    if prom > mayor:
        mayor = prom
        mejor = alumno

print(f"Mejor promedio: {mejor}")
```

```python
#ejercicio 9#
precios = {
    "playera": 350,
    "gorra": 220,
    "parche": 80,
    "sudadera": 650
}

carrito = ["playera", "gorra", "playera",
           "parche", "parche", "parche"]

cantidades = {}

for producto in carrito:
    if producto not in cantidades:
        cantidades[producto] = 0

    cantidades[producto] += 1

subtotal = 0

for producto, cantidad in cantidades.items():
    importe = cantidad * precios[producto]
    subtotal += importe
    print(f"{cantidad} x {producto} = ${importe}")

print(f"Subtotal: ${subtotal}")

if subtotal > 800:
    descuento = subtotal * 0.10
else:
    descuento = 0

total = subtotal - descuento

print(f"Descuento: ${descuento}")
print(f"Total: ${total}")
```

```python
#ejercicio 10#
abrev = {
    "q": "que",
    "xq": "porque",
    "tmb": "también",
    "pq": "porque",
    "bn": "bien",
    "tqm": "te quiero mucho"
}

def expandir(mensaje):
    palabras = mensaje.lower().split()
    traducidas = []

    for palabra in palabras:
        traducidas.append(abrev.get(palabra, palabra))

    return " ".join(traducidas)

print(expandir("Q onda tmb voy al concierto xq tocan bn"))
print(expandir("tqm Mamá"))
```

```python
#ejercicio 11#
def rendimiento():
    while True:
        try:
            km = float(input("Kilómetros recorridos: "))
            litros = float(input("Litros cargados: "))

            resultado = round(km / litros, 1)
            return resultado

        except ValueError:
            print("Escribe solo números")

        except ZeroDivisionError:
            print("Los litros no pueden ser 0")


resultado = rendimiento()

print(f"Rendimiento: {resultado} km/l")

if resultado >= 20:
    print("Excelente")
elif resultado >= 15:
    print("Normal")
else:
    print("Revisa tu moto")
```

```python
#ejercicio 12#
def precio_boleto(edad):
    if edad < 12:
        return 150
    elif edad < 60:
        return 300
    else:
        return 180


total = 0

for persona in range(1, 4):
    while True:
        try:
            edad = int(input(f"Edad de la persona {persona}: "))

            if edad < 0 or edad > 120:
                print("Edad fuera de rango")
                continue

            break

        except ValueError:
            print("Escribe la edad con número")

    precio = precio_boleto(edad)

    print(f"Boleto: ${precio}")
    total += precio

print(f"Total a pagar: ${total}")
```

```python
#ejercicio 14#
sala = [
    ["X", "L", "L"],
    ["L", "X", "L"],
    ["X", "X", "L"]
]

def reservar(sala):
    try:
        fila = int(input("Fila (0-2): "))
        columna = int(input("Columna (0-2): "))

        if sala[fila][columna] == "X":
            print("Ocupado, elige otro")
            return False

        sala[fila][columna] = "X"
        print("¡Reservado!")
        return True

    except ValueError:
        print("Escribe números")
        return False

    except IndexError:
        print("Ese asiento no existe")
        return False


while not reservar(sala):
    pass

for fila in sala:
    for asiento in fila:
        print(asiento, end=" ")
    print()
```

```python
#ejercicio 15#
cupones = {
    "METAL10": 10,
    "MOTO20": 20,
    "ROCK15": 15
}

def aplicar(total, codigo):
    codigo = codigo.strip().upper()

    try:
        porcentaje = cupones[codigo]
        return total - (total * porcentaje / 100)

    except KeyError:
        print("Cupón no válido")
        return total


codigos = [" moto20 ", "GRATIS", "metal10"]

for codigo in codigos:
    resultado = aplicar(1200, codigo)
    print(f"'{codigo}' → ${resultado}")
```

```python
#ejercicio 16#
def evaluar(pw):
    fallas = []

    if len(pw) < 8:
        fallas.append("mínimo 8 caracteres")

    tiene_numero = False
    tiene_mayuscula = False
    tiene_espacio = False

    for caracter in pw:
        if caracter.isdigit():
            tiene_numero = True

        if caracter.isupper():
            tiene_mayuscula = True

        if caracter == " ":
            tiene_espacio = True

    if not tiene_numero:
        fallas.append("falta un número")

    if not tiene_mayuscula:
        fallas.append("falta una mayúscula")

    if tiene_espacio:
        fallas.append("sin espacios")

    return fallas


contrasenas = ["metal", "Harley2008", "road glide 08", "KLR650moto"]

for contrasena in contrasenas:
    fallas = evaluar(contrasena)

    if len(fallas) == 0:
        print(f"{contrasena}: segura")
    else:
        print(f"{contrasena}: {fallas}")
```

```python
#ejercicio 17#
def es_palindromo(texto):
    limpio = texto.lower().replace(" ", "")
    return limpio == limpio[::-1]


frases = ("Anita lava la tina", "ABBA", "Metallica", "Oso")

for frase in frases:
    print(f"{frase}: {es_palindromo(frase)}")
```

```python
#ejercicio 18#
frase = "el rock nunca va a morir en esta ciudad"

def vocales(palabra):
    contador = 0

    for letra in palabra:
        if letra in "aeiou":
            contador += 1

    return contador


conteo = {}

for palabra in frase.split():
    conteo[palabra] = vocales(palabra)

mas_vocales = ""
mayor = -1

for palabra, cantidad in conteo.items():
    if cantidad > mayor:
        mayor = cantidad
        mas_vocales = palabra

print(conteo)
print(f"Más vocales: {mas_vocales} ({mayor})")
```

```python
#ejercicio 19#
alumnos = [
    "Bruce Dickinson",
    "Brian Dickinson",
    "Ana Lopez",
    "Andrea Lopez",
    "Angel Lopez"
]

usados = []

def usuario(nombre):
    partes = nombre.lower().split()

    base = partes[0][0] + partes[-1]

    candidato = base
    numero = 2

    while candidato in usados:
        candidato = base + str(numero)
        numero += 1

    usados.append(candidato)

    return candidato


for nombre in alumnos:
    print(f"{nombre} → {usuario(nombre)}")
```

```python
#ejercicio 20#
prohibidas = ["tonto", "menso", "baboso"]

def censurar(frase):
    palabras = frase.split()
    cantidad = 0

    for i in range(len(palabras)):
        if palabras[i].lower() in prohibidas:
            palabras[i] = "*" * len(palabras[i])
            cantidad += 1

    return (" ".join(palabras), cantidad)


frases = [
    "no seas menso y pásame el código",
    "Tonto el que no haga commit",
    "todo bien profe"
]

for frase in frases:
    texto, cantidad = censurar(frase)
    print(f"{texto}  ({cantidad} censuradas)")
```
