Descripción

El programa consiste en un menú modular desarrollado en Python que permite ejecutar cuatro secciones diferentes: tuplas, diccionarios, excepciones y strings. Cada sección cuenta con su propia función y demuestra el uso de los conceptos aprendidos durante la semana. El menú utiliza un ciclo `while` para mantenerse activo hasta que el usuario seleccione la opción de salir.

```python

def sumar_tupla(tupla):
    total = 0
    for numero in tupla:
        total = total + numero
    return total


def seccion_tuplas():
    print("\n========== SECCION 1: TUPLAS ==========")
    numeros = (10, 20, 30, 40, 50)
    print("Tupla original:", numeros)
    print("Tercer elemento:", numeros[2])
    numero1 = int(input("Ingresa el primer numero: "))
    numero2 = int(input("Ingresa el segundo numero: "))
    nueva_tupla = numeros + (numero1, numero2)
    print("Nueva tupla:", nueva_tupla)
    lista_numeros = list(nueva_tupla)
    lista_numeros.sort()
    print("Lista ordenada:", lista_numeros)
    resultado = sumar_tupla(nueva_tupla)
    print("Suma de la nueva tupla:", resultado)

contactos = {
    "Ana": "4421234567",
    "Luis": "4422345678",
    "Maria": "4423456789"}

def buscar_telefono(diccionario, nombre):
    return diccionario.get(nombre)

def seccion_diccionarios():
    print("\n========== SECCION 2: DICCIONARIOS ==========")
    nombre = input("Ingresa el nombre del nuevo contacto: ").strip()
    telefono = input("Ingresa su numero de telefono: ").strip()
    contactos[nombre] = telefono
    print("\nNombres registrados:")
    for nombre_contacto in contactos:
        print(nombre_contacto)
    nombre_buscar = input("\nQue contacto deseas buscar?: ").strip()
    telefono_encontrado = buscar_telefono(contactos, nombre_buscar)
    if telefono_encontrado is not None:
        print("Telefono de", nombre_buscar, ":", telefono_encontrado)
    else:
        print(nombre_buscar, "no esta en tus contactos")

def seccion_excepciones():
    print("\n========== SECCION 3: EXCEPCIONES ==========")
    try:
        numero1 = int(input("Ingresa el primer numero entero: "))
        numero2 = int(input("Ingresa el segundo numero entero: "))
        suma = numero1 + numero2
        print("La suma es:", suma)
        division = numero1 / numero2
        print("La division es:", division)
    except ValueError:
        print("Error: debes ingresar numeros enteros validos. No dejes campos vacios.")
    except ZeroDivisionError:
        print("Error: no puedes dividir entre cero. Intenta con otro divisor.")

def contar_palabras(texto):
    palabras = texto.split()
    return len(palabras)

def seccion_strings():
    print("\n========== SECCION 4: STRINGS ==========")
    mensaje = "La programacion permite resolver problemas reales"
    print("Mensaje original:", mensaje)
    print("Longitud del mensaje:", len(mensaje))
    print("Mensaje en mayusculas:", mensaje.upper())
    mensaje_modificado = mensaje.replace("programacion", "tecnologia")
    print("Mensaje reemplazado:", mensaje_modificado)
    cantidad_palabras = contar_palabras(mensaje)
    print("Cantidad de palabras:", cantidad_palabras)

def menu_principal():
    while True:
        print("\n================================")
        print("          MENU MODULAR")
        print("================================")
        print("1) Tuplas")
        print("2) Diccionarios")
        print("3) Excepciones")
        print("4) Strings")
        print("5) Salir")
        print("================================")
        opcion = input("Elige una opcion: ").strip()
        if opcion == "1":
            seccion_tuplas()
        elif opcion == "2":
            seccion_diccionarios()
        elif opcion == "3":
            seccion_excepciones()
        elif opcion == "4":
            seccion_strings()
        elif opcion == "5":
            print("\nPrograma finalizado. Hasta pronto!")
            break
        else:
            print("\nOpcion invalida. Selecciona una opcion del 1 al 5.")
menu_principal()
```
