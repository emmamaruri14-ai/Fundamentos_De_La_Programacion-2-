```python
PRECIO_TORTILLA = 27.00
PRECIO_ARROZ = 35.00
PRECIO_SALSA = 21.00
PRECIO_FRIJOLES = 40.00

cantidad_tickets = 0
dinero_ganado = 0.00

print("========================================")
print("       BIENVENIDO A LA TORTILLERIA")
print("========================================")
print("Caja registradora iniciada correctamente.")

while True:

    print("\n========================================")
    print("       NUEVA OPERACION")
    print("========================================")
    print("Escriba 'si' para iniciar un nuevo ticket.")
    print("Escriba 'fin' para terminar el dia.")

    comando = input("Seleccione una opcion: ").strip().lower()

    while comando not in ["si", "fin"]:
        print("Opcion no valida. Escriba si o fin.")
        comando = input("Seleccione una opcion: ").strip().lower()

    if comando == "fin":
        break

    compra = []
    print("\nMENU DE TORTILLAS")
    print("1. Medio kilo")
    print("2. 1 kilo")
    print("3. 2 kilos")
    print("4. 3 kilos")
    print("5. 4 kilos")

    opcion = input("Seleccione una opcion (1-5): ").strip()

    while opcion not in ["1", "2", "3", "4", "5"]:
        print("Opcion no valida.")
        opcion = input("Seleccione una opcion (1-5): ").strip()

    if opcion == "1":
        kilos = 0.5
    elif opcion == "2":
        kilos = 1.0
    elif opcion == "3":
        kilos = 2.0
    elif opcion == "4":
        kilos = 3.0
    else:
        kilos = 4.0

    subtotal = kilos * PRECIO_TORTILLA

    compra.append({
        "nombre": "Tortillas",
        "cantidad": kilos,
        "unidad": "kg",
        "precio": PRECIO_TORTILLA,
        "subtotal": subtotal
    })
    while True:

        respuesta = input(
            "\n¿Desea agregar otro producto? (si/no): "
        ).strip().lower()

        while respuesta not in ["si","no"]:
            print("Respuesta no valida.")
            respuesta = input(
                "Escriba si o no: "
            ).strip().lower()

        if respuesta == "no":
            break

        print("\n========== MENU DE PRODUCTOS ==========")
        print("1. Arroz     - $35.00")
        print("2. Salsa     - $21.00")
        print("3. Frijoles  - $40.00")

        producto = input("Seleccione un producto (1-3): ").strip()

        while producto not in ["1", "2", "3"]:
            print("Opcion no valida.")
            producto = input(
                "Seleccione un producto (1-3): "
            ).strip()

        if producto == "1":
            nombre = "Arroz"
            precio = PRECIO_ARROZ
        elif producto == "2":
            nombre = "Salsa"
            precio = PRECIO_SALSA
        else:
            nombre = "Frijoles"
            precio = PRECIO_FRIJOLES

        cantidad = input(
            f"¿Cuantas unidades de {nombre} desea?: "
        ).strip()

        while not cantidad.isdigit() or int(cantidad) <= 0:
            print("Ingrese un numero entero mayor que cero.")
            cantidad = input(
                f"¿Cuantas unidades de {nombre} desea?: "
            ).strip()

        cantidad = int(cantidad)
        subtotal = precio * cantidad

        compra.append({
            "nombre": nombre,
            "cantidad": cantidad,
            "unidad": "unidad(es)",
            "precio": precio,
            "subtotal": subtotal
        })

        print(f"{nombre} agregado correctamente.")

    print("\n========================================")
    print("             TORTILLERIA")
    print("             TICKET DE COMPRA")
    print("========================================")

    total = 0.00

    for producto in compra:

        print(f"\nProducto: {producto['nombre']}")
        print(
            f"Cantidad: {producto['cantidad']} "
            f"{producto['unidad']}"
        )
        print(f"Precio unitario: ${producto['precio']:.2f}")
        print(f"Subtotal: ${producto['subtotal']:.2f}")

        total += producto["subtotal"]

    print("\n========================================")
    print(f"TOTAL A PAGAR: ${total:.2f}")
    print("========================================")
    print("Gracias por su compra.")

    cantidad_tickets += 1
    dinero_ganado += total

    print("\nTicket registrado correctamente.")
    print(f"Tickets realizados: {cantidad_tickets}")

print("\n========================================")
print("          CIERRE DE CAJA")
print("========================================")

print(f"Cantidad de tickets: {cantidad_tickets}")
print(f"Dinero recaudado: ${dinero_ganado:.2f}")

print("----------------------------------------")

if cantidad_tickets == 0:
    print("No se realizaron ventas.")
else:
    print("Jornada de ventas finalizada correctamente.")

print("========================================")
print("       FIN DEL PROGRAMA")
print("========================================")
```
