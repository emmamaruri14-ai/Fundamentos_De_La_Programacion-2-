```python

# ==========================================
# PROTOTIPO: CAJA REGISTRADORA DE TORTILLERIA
# ==========================================

# 1. PRECIOS DE LOS PRODUCTOS
PRECIO_TORTILLA = 27.00
PRECIO_ARROZ = 35.00
PRECIO_SALSA = 21.00
PRECIO_FRIJOLES = 40.00

# 2. VARIABLES INICIALES
compra = []

print("========================================")
print("       BIENVENIDO A LA TORTILLERIA")
print("========================================")

# ------------------------------------------
# 3. SELECCION DE TORTILLAS
# ------------------------------------------

print("\nMENU DE TORTILLAS")
print("1. Medio kilo")
print("2. 1 kilo")
print("3. 2 kilos")
print("4. 3 kilos")
print("5. 4 kilos")

opcion = input("Seleccione una opcion (1-5): ").strip()

while opcion not in ["1", "2", "3", "4", "5"]:
    print("Error: seleccione una opcion del 1 al 5.")
    opcion = input("Seleccione una opcion (1-5): ").strip()

if opcion == "1":
    kilos = 0.5
elif opcion == "2":
    kilos = 1.0
elif opcion == "3":
    kilos = 2.0
elif opcion == "4":
    kilos = 3.0
else:
    kilos = 4.0

subtotal = kilos * PRECIO_TORTILLA

compra.append({
    "nombre": "Tortillas",
    "cantidad": kilos,
    "unidad": "kg",
    "precio": PRECIO_TORTILLA,
    "subtotal": subtotal
})

print(f"Subtotal de tortillas: ${subtotal:.2f}")

# ------------------------------------------
# 4. COMPRA DE PRODUCTOS ADICIONALES
# ------------------------------------------

while True:
    respuesta = input(
        "\n¿Desea agregar otro producto? (si/no): "
    ).strip().lower()

    while respuesta not in ["si", "sí", "no"]:
        print("Error: escriba si o no.")
        respuesta = input(
            "¿Desea agregar otro producto? (si/no): "
        ).strip().lower()

    if respuesta == "no":
        break

    print("\n========== MENU DE PRODUCTOS ==========")
    print("1. Arroz     - $35.00 por unidad")
    print("2. Salsa     - $21.00 por unidad")
    print("3. Frijoles  - $40.00 por unidad")

    producto = input("Seleccione un producto (1-3): ").strip()

    while producto not in ["1", "2", "3"]:
        print("Error: seleccione una opcion del 1 al 3.")
        producto = input(
            "Seleccione un producto (1-3): "
        ).strip()

    if producto == "1":
        nombre = "Arroz"
        precio = PRECIO_ARROZ
    elif producto == "2":
        nombre = "Salsa"
        precio = PRECIO_SALSA
    else:
        nombre = "Frijoles"
        precio = PRECIO_FRIJOLES

    cantidad = input(
        f"¿Cuantas unidades de {nombre} desea?: "
    ).strip()

    while not cantidad.isdigit() or int(cantidad) <= 0:
        print("Error: ingrese un entero mayor que cero.")
        cantidad = input(
            f"¿Cuantas unidades de {nombre} desea?: "
        ).strip()

    cantidad = int(cantidad)
    subtotal = precio * cantidad

    compra.append({
        "nombre": nombre,
        "cantidad": cantidad,
        "unidad": "unidad(es)",
        "precio": precio,
        "subtotal": subtotal
    })

    print(f"{nombre} agregado correctamente.")
    print(f"Subtotal del producto: ${subtotal:.2f}")


# ------------------------------------------
# 5. IMPRESION DEL TICKET
# ------------------------------------------

print("\n========================================")
print("          TORTILLERIA")
print("          TICKET DE COMPRA")
print("========================================")

total = 0.00

for producto in compra:
    print(f"\nProducto: {producto['nombre']}")
    print(f"Cantidad: {producto['cantidad']} {producto['unidad']}")
    print(f"Precio unitario: ${producto['precio']:.2f}")
    print(f"Subtotal: ${producto['subtotal']:.2f}")

    total += producto["subtotal"]

print("\n========================================")
print("RESUMEN DE PRODUCTOS")

for producto in compra:
    print(
        f"- {producto['nombre']}: "
        f"{producto['cantidad']} {producto['unidad']}"
    )

print("----------------------------------------")
print(f"TOTAL A PAGAR: ${total:.2f}")
print("========================================")
print("¡Gracias por su compra!")
print("¡Vuelva pronto!")
```
